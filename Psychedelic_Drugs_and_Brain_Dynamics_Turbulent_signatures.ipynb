### Framework


In order to determine the psychedelic-specific changes to the functional hierarchy, the authors framework employed two complementary model-free and model-based approaches inspired by the **turbulence theory of physics** ([Frisch, 1995](https://www.cambridge.org/highereducation/books/turbulence/FD8C5E35E5F1CA850E017461942A59AC#overview); [Kolmogorov, 1941](https://www.ams.jhu.edu/~eyink/Turbulence/classics/Kolmogorov41a.pdf); [Kuramoto, 1984](https://link.springer.com/chapter/10.1007/978-3-642-69689-3_7)).

- ***Model-free approach:*** we are going to measure information transmission across space and timescales using four novel and distinct measures recently introduced by [Deco and Kringelbach (2020)](https://pubmed.ncbi.nlm.nih.gov/33296654/): turbulence, information transfer, information cascade flow, and information cascade.
- ***Model-based approach:*** we are going to use a whole-brain model to mechanistically explain the statistical dependencies of the neuronal dynamics observed in the empirical neuroimaging data ([Deco, Cabral, et al., 2017](https://pubmed.ncbi.nlm.nih.gov/28315461/); [Deco & Kringelbach (2020)](https://pubmed.ncbi.nlm.nih.gov/33296654/); [Deco, et al., 2017](https://pubmed.ncbi.nlm.nih.gov/28596608/)). The model will allow us to study simulated perturbation-elicited changes in global and local brain activity to assess the system’s susceptibility and information-encoding capability, offering new insights into the brain’s reactivity to external perturbations, such as perturbations induced pharmacologically or by electrical stimulations.


### Model-Free Framework

Lewis F. Richardson observed a fascinating phenomenon in fluid dynamics: the formation of wheels or vortices of different sizes, creating a spatial hierarchical structure that can efficiently transmit energy and information.

![alt](https://www.lpp.polytechnique.fr/local/cache-vignettes/L630xH360/une_20200203_2-767f5.jpg?1679559490)

These wheels and vortices form a cascade, an ordered hierarchy based on their size across spatial scales. Kolmogorov interpreted this structure as an efficient mechanism for energy transmission through space. He introduced the concept of structure functions, which can be defined as:

\begin{equation}
S(r) = \left\langle \left( u(\bar{x} + r) - u(\bar{x}) \right)^2 \right\rangle = 2 \left[ B(0) - B(r) \right]
\end{equation} <br/>

The structure function represents the quadratic distance of displacement over a distance $r$ of the main variance. In Kolmogorov's case, the variance was velocity, while in our context, it is the BOLD signal. Interestingly, we can separate this into two terms, with the main term being $B(r)$. $B(r)$ is essentially the correlation of two signals located at points separated by an Euclidean distance $r$. Mathematically, it is expressed as:

\begin{equation}
B(r) = \left\langle u(\bar{x} + r )u(\bar{x}) \right\rangle
\end{equation} <br/>

where $u$ contains the spatio-temporal dynamics. $B(r)$ is analogous to what is known as functional connectivity in neuroimaging. We will examine functional connectivity as a function of the distance $r$ between the pairs of points we are observing.

Kolmogorov demonstrated both experimentally and theoretically that functional connectivity, when viewed as a function of space, follows a power law. This power law quantifies the intuition of Richardson and Kolmogorov: the hierarchical structure of wheels and vortices across space is remarkably efficient for transmitting energy and information (in the form of correlations). This hierarchical organization enhances the efficiency of energy and information transfer across spatial scales.

The matrix `rr`, sized NPARCELLS x NPARCELLS, store the Euclidean distances between pairs of brain regions. This distance computation is based on the Centers of Gravity (COGs) derived from the Schaefer atlas.

We calculated the maximum distance (`range_val`) among all pairs of COGs. This value is then divided into NR segments to establish a step size (`delta`) that will be used for subsequent analyses.

The `xrange` array is generated to represent a range of distances between brain regions, starting from half the step size (`delta / 2`) up to the maximum distance (`range_val`). This array facilitates the systematic exploration of connectivity patterns across different spatial scales.

Next, we define parameters and compute matrices essential for modeling connectivity dynamics across brain regions.

We start by defining LAMBDA, an array containing predefined decay factors $[0.21, 0.18, 0.15, 0.12, 0.09, 0.06, 0.03, 0.01]$. These factors determine the exponential decay rates used in subsequent calculations.

Next, `NLAMBDA` is computed to represent the number of elements in `LAMBDA`, providing the total number of decay rates considered.

A three-dimensional matrix `C1`is initialized with dimensions (`NLAMBDA, NPARCELLS, NPARCELLS`). This matrix will store connectivity coefficients computed based on the exponential decay of distances (`rr`) between pairs of brain regions.

For each decay factor `_lambda` in `LAMBDA`, we compute connectivity coefficients  `𝐶1[𝑖𝑙𝑎𝑚−1,𝑖,𝑗]`   using the formula ` 𝑛𝑝.𝑒𝑥𝑝(−𝜆∗𝑟𝑟[𝑖,𝑗])`. This computation assigns higher coefficients to regions closer in proximity `(lower rr[i, j] values)` and lower coefficients to regions farther apart.

This process facilitates the modeling of connectivity dynamics, capturing how the strength of connections between brain regions varies with distance, guided by the predefined decay factors in `LAMBDA`.

```python
C1 = np.zeros((NLAMBDA, NPARCELLS, NPARCELLS))

# Calculate C1
ilam = 0
for _lamba in LAMBDA:
    ilam += 1
    for i in range(NPARCELLS):
        for j in range(NPARCELLS):
            C1[ilam-1, i, j] = np.exp(-_lamba * rr[i, j])
```

Next we are going to process our empirical BOLD signal in few steps:

- Detrending: Each time-series `(ts[seed, :])` is detrended to remove any linear trends and then centered around zero by subtracting its mean.

- Signal Filtering: Using a Butterworth bandpass filter (`filtfilt`), we filter the detrended time-series data `(ts[seed, :])` to isolate frequency components of interest, ensuring that the filtered signal `(signal_filt[seed, :])` retains phase information crucial for further analysis.

- Analytic Signal and Phase Extraction: The Hilbert transform (`hilbert`) is applied to the filtered signal `(signal_filt[seed, :])` to obtain its analytic representation (`Xanalytic`). From this, instantaneous phase values `(Phases[seed, :])` are derived using `np.angle(Xanalytic)`, providing insights into the temporal dynamics of neuronal oscillations.

```python
signal_filt = np.zeros((NPARCELLS, Tmax))
Phases = np.zeros((NPARCELLS, Tmax))
Phases_su = np.zeros((NPARCELLS, Tmax))

sub = 0  # Set sub to 0

ts = data[sub]  # Get the time series data for subject 'sub'

for seed in range(NPARCELLS):
    ts[seed, :] = detrend(ts[seed, :] - np.mean(ts[seed, :]))
    signal_filt[seed, :] = filtfilt(bfilt, afilt, ts[seed, :])

    Xanalytic = hilbert(signal_filt[seed, :] - np.mean(signal_filt[seed, :]))
    Phases[seed, :] = np.angle(Xanalytic)
    rand_indices = np.random.permutation(Tmax)  # Generate random permutation indices
    Phases_su[seed, :] = Phases[seed, rand_indices]
```

Next we are going to measure information transmission across space and time- scales using four novel and distinct measures recently introduced by [Deco and Kringelbach (2020)](https://pubmed.ncbi.nlm.nih.gov/33296654/): turbulence, information transfer, information cascade flow, and information cascade.


### **Turbulence**

The amplitude turbulence, $R_{\lambda}(\overline{x}, t)$, is defined as the modulus of the Kuramoto local order parameter for a given brain node as a function of time:  <br/>
 <br/>
\begin{equation}
   R_{\lambda}(\overline{x}, t) =
    \left\lVert k \int_{-\infty}^{\infty}  d\overline{x}^{\prime} G_{\lambda}  (\overline{x}-\overline{x}^{\prime}) e^{i\varphi(\overline{x}^{\prime}, t)} \right\rVert
\end{equation} <br/>
 <br/>
where:

- $\left\lVert \right\rVert$ is the modulus of the complex number
- $G_{\lambda} $ is the local weighting kernel
- $G_{\lambda} (\overline{x}) = e^{-\lambda | \overline{x}|}$, $\varphi(\overline{x}, t)$ are the phases of the spatiotemporal data
- $k$ is the normalization factor
$[\int_{-\infty}^{\infty} d\overline{x}' G_{\lambda} (\overline{x} - x')]^{-1}$
- $\lambda$ defines the spatial scaling.

Hence, $R_{\lambda}$ represents the local levels of synchronization at a given scale, $\lambda$, as function of space, $\overline{x}$, and time, $t$. Inspired by the rotational vortices found in fluid dynamics, the turbulence measure characterizes the brain vortex space,  $R_{\lambda}$, over time.

The level of amplitude turbulence, $D_\lambda$, is defined as the standard deviation across time and space of the modulus of local Kuramoto order parameter $R_\lambda$.

\begin{equation}
D_y = \left\langle R_\lambda^2 \right\rangle_{\overline{x},t} - ((R_\lambda)^2) _{\overline{x},t^2}
\end{equation} <br/>


where the brackets $\left\langle  \right\rangle_{\overline{x},t}$ denote averages across time and space.

The variable name Enstrophy is because it is the concept from fluid dynamics and turbulence theory that quantifies the intensity of vorticity (the tendency of a fluid or system to rotate). It is often used to measure the level of small-scale turbulent activity in a system.




```python
enstrophy = np.zeros((NLAMBDA, NPARCELLS, Tmax))
Turbulence_node_sub = np.zeros((NLAMBDA, NPARCELLS))
enstrophy_su = np.zeros((NLAMBDA, NPARCELLS, Tmax))

for i in range(NPARCELLS):
  for ilam in range(NLAMBDA):
    C1lam=C1[ilam,:,:]
    sumphases = np.nansum(np.tile(C1lam[i, :], (Tmax, 1)).T * (np.cos(Phases) + 1j * np.sin(Phases)), axis=0) / np.nansum(C1lam[i, :])
    enstrophy[ilam, i, :] = np.abs(sumphases)
    Turbulence_node_sub[ilam, i] = np.nanstd(enstrophy[ilam, i, :])


    # Compute sum of phases for surrogate phases
    sumphases_su = np.nansum(np.tile(C1lam[i, :], (Tmax, 1)).T * (np.cos(Phases_su) + 1j * np.sin(Phases_su)), axis=0) / np.nansum(C1lam[i, :])
    enstrophy_su[ilam, i, :] = np.abs(sumphases_su)
```


### **Information transfer**

. The spatial information transfer shows how the information travels across space at a specific scale, λ. This measure is defined as the slope of a linear fitting in log-log scale of the time correlation between the Kuramoto local order parameter of two brain areas, at the same scale as a function of its Euclidean distance (r) within the inertial subrange (the limited range where turbulence energy is transferred from larger to smaller scales without loss). We used the linear fit only to quantify the level of decay of the correlation of the local level of synchronization with distance. Please note that we compute the linear fit in a limited log-log space and, as such, we do not have a distribution, and thus are not trying to fit a power law.

\begin{equation}
log(corr_t(R_\lambda(\overline{x}), R_\lambda(\overline{x} + r))) = A_\lambda \ x \log(r) + B\lambda
\end{equation} <br/>

where:

- $corr_t$ is the pairwise correlation across time,
- $A_\lambda$ and $B_\lambda$ are the fitting parameters for each scale (λ), where r is the spatial distance in the brain. The negative slope (Aλ) stands for the transfer in the spatial direction r of the information in terms of time correlation of the local level of synchronization.

In this regard, when the slope is steeper, the information travels across shorter distances, while a flatter slope indicates that the information is transferred across longer distances. Thus, the negative slope stands for the spatial information transfer. Note that the parameter A depends on only λ. It depends neither on t as the correlation is over time, nor on the brain areas, as the pairwise correlations are organized as a function of the Euclidean distances.

### **Information cascade flow**

 The information cascade flow characterizes the stream of information between a given scale $(λ)$ and a subsequent lower scale (λ − Δλ, where Δλ is a scale step) in consecutive time steps (t and t + Δt). In this way, the information cascade flow covers the information transfer across scales computed as the time correlation between the Kuramoto local order parameter in two consecutive scales and times:

\begin{equation}
F_y = \left\langle corr_t(R_\lambda) \overline{x},t +  Δt), R_\lambda - _{Δ,\lambda} (\overline{x},t)) \right\rangle_{\overline{x}}
\end{equation} <br/>


where the brackets $\left\langle  \right\rangle$ denote averages across time and space, and corrt refers to the pair-wise time correlations.

### **Information cascade**

Finally, the information cascade is defined by averaging the information transfer across scales λ, capturing the entire information-processing behavior across scales.

### Model-based Framework

Now that we have found an effect on the empirical data we are ready to dig into the inner mechanism of it using a Model-based approach.
We will build a whole-brain computational model where the local dynamics of each brain area (node) are described by the normal form of a supercritical Hopf bifurcation (also known as [Stuart-Landau](https://en.wikipedia.org/wiki/Stuart%E2%80%93Landau_equation)).

**Hopf Bifurcation:**

A bifurcation is a change in the behavior of a system as a parameter is varied. A Hopf bifurcation occurs when a system changes from a stable state (no oscillations) to oscillatory behavior (regular cycles) as a parameter crosses a critical value. Supercritical means this transition is smooth and continuous, leading to stable oscillations.

**Stuart-Landau Model:**

This is a specific form of the equations used to describe the Hopf bifurcation. It includes terms that account for both the growth of oscillations and the stabilization of these oscillations at a certain amplitude. Each node can exhibit different behaviors based on its parameters:
- Stable State (No Oscillations): If the parameter $a_n$ is negative, the brain area remains in a noisy, non-oscillatory state.
- Oscillatory State (Sustained Oscillations): If $a_n$ is positive, the brain area shows regular, sustained oscillations.
- Critical State: If $a_n$ is around zero, the brain area is at the point of transitioning between noise and oscillations.

When the parameter $a_n$ is slightly negative or close to zero, the brain area may exhibit oscillations that are influenced by random fluctuations (noise). When $a_n$ is positive, the brain area exhibits stable, regular oscillations that persist over time.

**In a nutshell**

The Stuart-Landau Model is a specific type of mathematical model used to describe how systems can transition from stable behavior (no oscillations) to periodic oscillations (regular, sustained cycles). In the context of the brain, it models how a brain area can switch from random, noisy activity to regular, rhythmic activity [Deco, Cabral, et al., 2017](https://www.ncbi.nlm.nih.gov/pmc/articles/PMC5440176/).

\begin{equation}
\frac{dx_n}{dt} = a_n x_n - \left[ x_n^2 + y_n^2 \right] x_n - \omega_n y_n + \nu \eta_n(t); \quad (1)
\end{equation}

\begin{equation}
\frac{dy_n}{dt} = a_n y_n - \left[ x_n^2 + y_n^2 \right] y_n + \omega_n x_n + \nu \eta_n(t); \quad (2)
\end{equation}

The equations you provided describe the dynamics of a whole-brain computational model where each brain area (node) is represented by a pair of variables ($x_n$, $y_n$).

**Where:**
- $x_n$ and $y_n$: Variables representing the state variables (e.g., neural activity) of brain area $n$.
- $a_n$: Local bifurcation parameter governing the dynamics of brain area $n$. It determines whether the area exhibits stable fixed points $(a_n<0)$ stable limit cycle oscillations $(a_n>0)$ with frequency $f_{n} = \omega_n/2π$, or a critical behavior $(a_n∼0)$
- $\left[ x_n^2 + y_n^2 \right] x_n$ and $\left[ x_n^2 + y_n^2 \right] y_n$: Nonlinear terms inducing oscillatory behavior.
- $\omega_n$: Frequency of oscillation of brain area $n$, from empirical data.
- $\nu \eta_n(t)$: Additive Gaussian noise term with standard deviation  $\nu = 0.01$, representing stochastic fluctuations.
- $\eta_n(t)$: Gaussian white noise process.





```python
# Define the local dynamics equations
def local_dynamics(state, t, a, omega):
    x, y = state
    dxdt = a * x - (x**2 + y**2) * x - omega * y
    dydt = a * y - (x**2 + y**2) * y + omega * x
    return [dxdt, dydt]

# Parameters
a = -0.1  # Negative value for stable fixed point
omega = 2 * np.pi * 1  # 1 Hz frequency (this parameter is less important when a < 0)
initial_state = [1, 0]
t = np.linspace(0, 1000, 10000)

# Solve the ODEs
solution = odeint(local_dynamics, initial_state, t, args=(a, omega))

# Plot the results
plt.figure(figsize=(10, 5))
plt.plot(t[:300], solution[:300, 0], label='x(t)')
plt.plot(t[:300], solution[:300, 1], label='y(t)')
plt.xlabel('Time')
plt.ylabel('Activity')
plt.title('Local Dynamics of a Single Node')
plt.legend()
plt.show()
```

Now that we have seen the local dynamics, we are ready to insert them into a parcellations in order to build the whole-brain dynamics which can be described by the following set of coupled equation:

\begin{equation}
\frac{dx_n}{dt} = a_n x_n - \left[ x_n^2 + y_n^2 \right] x_n - \omega_n y_n + G\sum_{p=1}^{N} C_{np}(x_p(t) - x_n) + \nu \eta_n(t); \quad (3)
\end{equation}
\begin{equation}
\frac{dy_n}{dt} = a_n y_n - \left[ x_n^2 + y_n^2 \right] y_n + \omega_n x_n + G\sum_{p=1}^{N} C_{np}(y_p(t) - y_n) + \nu \eta_n(t); \quad (4)
\end{equation}

In addition to the terms in equations (1) and (2):

- $G$: Represents the global conductivity of the fibers, scaling the structural connectivity between brain areas.
- $C_{np}$: Elements of the anatomical matrix that describe the connectivity strength between brain areas
- $x_{p}(t)$ and $y_{p}(t)$: State variables of brain area $p$ influencing brain area $n$ through connectivity.
- $N$: Total number of brain areas considered in the model.

These coupled differential equations (3) and (4) describe how each brain area's dynamics $(x_n, y_n)$ are influenced by its own internal dynamics, noise, and the activity of connected brain areas, mediated by the connectivity matrix $C_{np}$

This whole-brain computational model captures the transition from noise-induced activity to sustained oscillations, simulating the complex dynamics observed in the brain.

In this section, we define and execute a simulation loop using the `numba` library to efficiently compute the dynamic evolution of neuronal activities in a brain network model.

We define a `simulation_loop` using the `@njit` decorator from numba, which optimizes the function for just-in-time compilation and execution.

The function initializes variables required for simulation, including `wC`, `z`, `sumC`,`a`, `omea`, `dsig`, `dt`, and `Tmax`.
Loop Execution: Inside the loop, neuronal activities `z` are updated iteratively using the model equations, incorporating factors such as connectivity strengths `wC`, intrinsic dynamics `a`, external stimuli `omega`, noise `dsig`, and time step `dt`.
After completing the simulation loop, the updated neuronal activities z are returned.


Following the simulation loop, another function is defined to extract time-series data
The function initializes `nn`, and  `NPARCELLS`, essential for tracking and managing time-series data `xs` and the number of brain regions.
Within the loop, neuronal activities `z` are updated similarly to the first function, incorporating connectivity strengths, intrinsic dynamics, external stimuli, noise, and time step.
Time-Series Extraction: Time-series data xs is recorded at intervals defined by `TR`, ensuring snapshots of neuronal activity are captured periodically.
After completing the loop, both the updated neuronal activities `z` and the extracted time-series data xs are returned for further analysis.


```python
@njit
def simulation_loop(wC, z, sumC, a, omega, dsig, dt, Tmax):
    for t in np.arange(0, Tmax, dt):
      suma = np.dot(wC, z) - (sumC * z)  # sum(Cij*xi) - sum(Cij)*xj
      zz = z[:, ::-1]  # Flipped z, because (x.*x + y.*y)
      z = z + dt * ((a * z) + (zz * omega) - (z * (z * z + zz * zz)) + suma) + (dsig * np.random.randn(NPARCELLS, 2))
    return z

# Run the simulation
z = simulation_loop(wC, z, sumC, a, omega, dsig, dt, Tmax)


@njit
def simulation_loop(wC, z, sumC, a, omega, dsig, dt, Tmax, TR, xs):
    nn = 0
    NPARCELLS = wC.shape[0]
    for t in np.arange(0, (Tmax - 1) * TR, dt):
        suma = np.dot(wC, z) - (sumC * z)  # sum(Cij*xi) - sum(Cij)*xj
        zz = z[:, ::-1]  # Flipped z, because (x.*x + y.*y)
        z = z + dt * (a * z + zz * omega - z * (z * z + zz * zz) + suma) + dsig * np.random.randn(NPARCELLS, 2)
        if abs(t % TR) < 0.01:
            xs[nn, :] = z[:, 0]
            nn += 1
    return z, xs


# Run the simulation
z, xs = simulation_loop(wC, z, sumC, a, omega, dsig, dt, Tmax, TR, xs)

ts = xs.T;
```

Now we are goingn to process our simulated signal in few steps:

- Detrending: Each time-series `(ts[seed, :])` is detrended to remove any linear trends and then centered around zero by subtracting its mean.

- Signal Filtering: Using a Butterworth bandpass filter (bfilt, afilt), we filter the detrended time-series data `(ts[seed, :])` to isolate frequency components of interest, ensuring that the filtered signal `(signal_filt[seed, :])` retains phase information crucial for further analysis.

- Analytic Signal and Phase Extraction: The Hilbert transform (hilbert) is applied to the filtered signal `(signal_filt[seed, :])` to obtain its analytic representation (Xanalytic). From this, instantaneous phase values `(Phases[seed, :])` are derived using np.angle(Xanalytic), providing insights into the temporal dynamics of neuronal oscillations.

```python
signal_filt = np.zeros((NPARCELLS, Tmax))
Phases = np.zeros((NPARCELLS, Tmax))

for seed in range(NPARCELLS):
    ts[seed, :] = detrend(ts[seed, :] - np.mean(ts[seed, :]))
    signal_filt[seed, :] = filtfilt(bfilt, afilt, ts[seed, :])

    Xanalytic = hilbert(signal_filt[seed, :] - np.mean(signal_filt[seed, :]))
    Phases[seed, :] = np.angle(Xanalytic)
```
